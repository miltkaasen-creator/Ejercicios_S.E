EJERCICIO 01 - a

```python
import numpy as np
import matplotlib.pyplot as plt
from fractions import Fraction

# Matrices
x = np.matrix([[5, 2],
               [-3, 3]])
y = np.matrix([[1],
               [5]])

# Resolver
resultado = np.linalg.solve(x, y)

# Extraer valores numéricos
x_sol, y_sol = resultado[0,0], resultado[1,0]

# Convertir a fracción
x_frac = Fraction(x_sol).limit_denominator()
y_frac = Fraction(y_sol).limit_denominator()

print("Solución en decimal:")
print(f"x = {round(x_sol,2)}, y = {round(y_sol,2)}")
print("")
print("Solución en fracción:")
print(f"x = {x_frac}, y = {y_frac}")

# Graficar las dos rectas
xx = np.linspace(-10, 10, 100)

# Ecuaciones despejadas
y1 = (1 - 5*xx)/2
y2 = (5 + 3*xx)/3

plt.plot(xx, y1, label="5x + 2y = 1", color="lightblue")
plt.plot(xx, y2, label="-3x + 3y = 5", color="thistle")

# Punto de intersección
plt.scatter(x_sol, y_sol, color="black", s=20, zorder=5, label="Solución")

plt.xlabel("Eje x")
plt.ylabel("Eje y")
plt.title("Sistema de Ecuaciones-Ejercicio 01-a", size=10, pad=20)
plt.legend()
plt.grid(True, linestyle="--", alpha=0.7)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Matrices
A = np.array([[-1, 2],
              [2, -4]], dtype=float)
B = np.array([4, 3], dtype=float)

# Comprobar determinante
det_A = np.linalg.det(A)
print("Determinante del sistema:", det_A)

if det_A == 0:
    print("El sistema no tiene solución única. Se puede graficar como rectas que no se intersectan exactamente.")
else:
    sol = np.linalg.solve(A, B)
    x_sol, y_sol = sol[0], sol[1]
    print("Solución: x =", x_sol, ", y =", y_sol)

# Crear rango para graficar
xx = np.linspace(-10, 10, 400)

# Rectas
y1 = (4 + xx)/2
y2 = (2*xx - 3)/4

# Graficar
plt.figure(figsize=(8,6))
plt.plot(xx, y1, color="green", label="-x + 2y = 4")
plt.plot(xx, y2, color="orange", label="2x - 4y = 3")

if det_A != 0:
    plt.scatter(x_sol, y_sol, color="black", s=80, zorder=5, label="Solución")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Sistema de Ecuaciones - Ejercicio 6-b", pad=20, size=10)
plt.legend()
plt.grid(True, linestyle="--", alpha=0.7)
plt.show()
```

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

#SymPy
a, b = sp.symbols('a b', real=True)

eq1 = sp.Eq(a + b, 122)
eq2 = sp.Eq( (a + b) + (a + 4), 180 )

sol_sym = sp.solve((eq1, eq2), (a, b), dict=True)
sol_sym = sol_sym[0]
a_sym = float(sol_sym[a])
b_sym = float(sol_sym[b])
c_sym = a_sym + 4.0

print("=== Problema 14 (SymPy) ===")
print(f"a (menor) = {a_sym:.0f}°")
print(f"b (mayor) = {b_sym:.0f}°")
print(f"c (tercero) = {c_sym:.0f}°")
print()

#NumPy
A = np.array([[1.0, 1.0],
              [2.0, 1.0]], dtype=float)
B = np.array([122.0, 176.0], dtype=float)

sol_np = np.linalg.solve(A, B)
a_np, b_np = sol_np
c_np = a_np + 4.0

print("=== Problema 14 (NumPy) ===")
print(f"a (menor) = {a_np:.0f}°")
print(f"b (mayor) = {b_np:.0f}°")
print(f"c (tercero) = {c_np:.0f}°")
print()

#Gráfico
angles = [a_np, b_np, c_np]
labels = [f"a = {int(a_np)}° (menor)", f"b = {int(b_np)}° (mayor)", f"c = {int(c_np)}° (tercero)"]
colors = ['pink', 'salmon', 'lightblue']

plt.figure(figsize=(6,5))
wedges, texts = plt.pie(angles, colors=colors, startangle=90, wedgeprops=dict(edgecolor='k'))
# Etiquetas
plt.legend(wedges, labels, title="Ángulos", loc="center left", bbox_to_anchor=(1, 0.5))
plt.title("Problema 14", pad=20, size=10)
plt.gca().set_aspect('equal')
plt.show()
```

Problema N°15

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Variables
x, y = sp.symbols('x y', real=True)

# Ecuaciones
eq1 = sp.Eq(x + y, 10000)
eq2 = sp.Eq(0.05*x, 0.035*y + 300)

# SymPy
sol_sym = sp.solve((eq1, eq2), (x, y), dict=True)[0]
x_sym, y_sym = sol_sym[x], sol_sym[y]

print("=== Problema 15 (SymPy) ===")
print(f"Inversión en producto 1 = {x_sym:.2f} €")
print(f"Inversión en producto 2 = {y_sym:.2f} €\n")

# NumPy
A = np.array([[1, 1],
              [0.05, -0.035]], dtype=float)
B = np.array([10000, 300], dtype=float)

sol_np = np.linalg.solve(A, B)
x_np, y_np = sol_np

print("=== Problema 15 (NumPy) ===")
print(f"Inversión en producto 1 = {x_np:.2f} €")
print(f"Inversión en producto 2 = {y_np:.2f} €\n")

# Gráfico
labels = ["Producto 1 (5%)", "Producto 2 (3.5%)"]
values = [x_np, y_np]
colors = ['lightpink', 'lightcoral']  # Colores pasteles

fig, ax = plt.subplots(figsize=(7,5))
bars = ax.bar(labels, values, color=colors, edgecolor='k', width=0.5)

# Etiquetas sobre las barras
for bar, val in zip(bars, values):
    ax.text(bar.get_x() + bar.get_width()/2, val + 100, f"{val:.0f} €",
            ha='center', va='bottom', fontsize=11, fontweight='bold')

# Limites seguros del eje Y
y_max_limit = max(values) + 500
ax.set_ylim(0, y_max_limit)

ax.set_ylabel("Inversión (€)")
ax.set_title("Problema 15 — Distribución de inversión", pad=20, size=10)
ax.grid(axis="y", linestyle="--", alpha=0.6)

plt.show()
```
